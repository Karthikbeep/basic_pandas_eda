```python
import numpy as np
import pandas as pd
```

```python
sales = pd.DataFrame({
    "OrderID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Date": [
        "2026-01-05", "2026-01-12", "2026-01-18", "2026-02-03",
        "2026-02-15", "2026-02-20", "2026-03-05", "2026-03-18"
    ],
    "Product": [
        "Laptop", "Phone", "Laptop", "Tablet",
        "Phone", "Laptop", "Tablet", "Phone"
    ],
    "Category": [
        "Electronics", "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Electronics", "Electronics"
    ],
    "City": [
        "Hyderabad", "Delhi", "Hyderabad", "Mumbai",
        "Delhi", "Mumbai", "Hyderabad", "Delhi"
    ],
    "Sales": [60000, 30000, 55000, 20000, 35000, 65000, 22000, 32000],
    "Quantity": [1, 2, 1, 1, 2, 1, 1, 2]
})
```

**Data Inspection**

```python
#checking number of rows and columns
sales.shape
```

```python
#checking the data types
sales.dtypes
```

```python
#checking the data quality
sales.isna().sum()
```

**Product Analysis**

```python
#total revenue by product, highest → lowest.
sales.groupby("Product")["Sales"].sum().sort_values(ascending = False)
```

**City Analysis**

```python
#total revenue by city, highest → lowest.
sales.groupby("City")["Sales"].sum().sort_values(ascending = False)
```

**Quantity**

```python
#total quantity sold by product.
sales.groupby("Product")["Quantity"].sum()
```

```python
#Find the overall average Sales.
avg_sales = sales['Sales'].mean()
avg_sales
```

```python
#Which product:generated the most revenue?
sales.groupby("Product")["Sales"].sum()
```
